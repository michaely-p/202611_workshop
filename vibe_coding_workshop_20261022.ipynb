````python
# OUTPUT CODE:

import matplotlib.pyplot as plt
import seaborn as sns

# Reshape the dataframe
heatmap_data = df.pivot_table(
    index="Name",                       # y-axis: university
    columns="Publication Year",         # x-axis: year
    values="Web of Science Documents",  # cell color/value
    aggfunc="sum"
)

# Optional: sort universities alphabetically
heatmap_data = heatmap_data.sort_index()

# Plot
plt.figure(figsize=(14, max(6, 0.5 * len(heatmap_data))))

sns.heatmap(
    heatmap_data,
    cmap="YlGnBu",
    annot=True,
    fmt=".0f",
    linewidths=0.5,
    cbar_kws={"label": "Web of Science Documents"}
)

plt.xlabel("Publication Year")
plt.ylabel("University")
plt.title("Web of Science Documents by University and Publication Year")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python
# OUTPUT CODE:

import os
import re
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
os.makedirs("university_lineplots", exist_ok=True)

for university, university_df in df.groupby("Name"):

    university_df = university_df.sort_values("Publication Year")

    fig, ax = plt.subplots(figsize=(9, 5))

    sns.lineplot(
        data=university_df,
        x="Publication Year",
        y="Web of Science Documents",
        marker="o",
        linewidth=2.5,
        errorbar=None,
        ax=ax
    )

    ax.set_title(f"Web of Science Documents — {university}")
    ax.set_xlabel("Publication Year")
    ax.set_ylabel("Web of Science Documents")
    ax.tick_params(axis="x", rotation=45)
    ax.grid(True, linestyle="--", alpha=0.5)

    # Create a filename safe for most operating systems
    safe_name = re.sub(r'[\\/*?:"<>|]', "_", university)

    fig.tight_layout()
    fig.savefig(
        f"university_lineplots/{safe_name}.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close(fig)
````
