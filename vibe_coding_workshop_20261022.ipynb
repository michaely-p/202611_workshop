````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Read the Excel file into a pandas DataFrame
# Replace 'your_data.xlsx' with your actual file path
file_path = "https://github.com/michaely-p/202611_workshop/raw/refs/heads/main/20260629_pseudo_dataset.xlsx"
df = pd.read_excel(file_path)

# 2. Set the plot style for better aesthetics
sns.set_theme(style="whitegrid")

# 3. Create the figure and plot the data
plt.figure(figsize=(10, 6))
sns.lineplot(
    data=df,
    x='Year',
    y='Publications',
    hue='Name',
    marker='o',          # Adds dots at each data point
    linewidth=2          # Makes the lines slightly thicker
)

# 4. Customize the title and axis labels
plt.title('University Publications Over Time', fontsize=16, pad=15)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Publications', fontsize=12)

# 5. Format the X-axis to show only whole years (since Year is an int64)
plt.xticks(df['Year'].unique())

# 6. Adjust the legend to sit outside the plot area
plt.legend(title='University Name', bbox_to_anchor=(1.05, 1), loc='upper left')

# 7. Adjust layout to prevent clipping of the legend
# plt.tight_layout()

# 8. Display the plot
plt.show()
````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Read the Excel file
# Replace 'your_data.xlsx' with your actual file path
file_path = "https://github.com/michaely-p/202611_workshop/raw/refs/heads/main/20260629_pseudo_dataset.xlsx"
df = pd.read_excel(file_path)

# 2. Pivot the DataFrame to create a matrix for the heatmap
# Rows (y) = Name, Columns (x) = Year, Values (color) = Publications
heatmap_data = df.pivot(index='Name', columns='Year', values='Publications')

# 3. Set up the matplotlib figure
plt.figure(figsize=(12, 6))

# 4. Draw the heatmap
sns.heatmap(
    heatmap_data,
    annot=True,       # Show the actual numbers in the boxes
    fmt=".0f",        # Format the numbers as integers
)

# 5. Customize titles and labels
plt.title('Heatmap of Publications by University and Year', fontsize=16, pad=15)
plt.xlabel('Year', fontsize=12)
plt.ylabel('University Name', fontsize=10)

# 6. Adjust layout to fit labels nicely
plt.tight_layout()

# 7. Display the plot
plt.show()
````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming your DataFrame is already loaded into 'df'
# df = pd.read_excel('your_file.xlsx')

# Set the aesthetic style of the plots
sns.set_theme(style="whitegrid")

# Get a list of all unique universities in the 'Name' column
universities = df['Name'].unique()

# Loop through each university
for uni in universities:
    # Filter the data for the current university
    subset_df = df[df['Name'] == uni]

    # Create a new figure for each plot
    plt.figure(figsize=(8, 5))

    # Plot the lineplot for the filtered data
    sns.lineplot(
        data=subset_df,
        x='Year',
        y='Publications',
        marker='o',      # Adds dots to the data points
        linewidth=2
    )

    # Customize the title and labels
    plt.title(f'Publications Over Time: {uni}', fontsize=14, pad=10)
    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Total Publications', fontsize=12)

    # Force the X-axis to display only whole years
    plt.xticks(subset_df['Year'].unique())

    # Adjust layout
    plt.tight_layout()

    # Display the plot
    plt.show()
````
